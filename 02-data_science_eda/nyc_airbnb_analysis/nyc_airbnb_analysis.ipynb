Después de ver de forma rápida el df lo primero que podemos hacer es cambiar los nombres de las carácterísticas para que sigan el mismo patrón para todas

```python

df = df.rename(columns={"PETS Allowed": "pets_allowed","swimming pool":"swimming_pool"})
```

Voy a borrar los - que haya (los - de week_price y price son de las mismas filas con borrar las filas donde week_price tenga - ya es suficiente, aún así lo mostraré luego para comprobarlo)

```python
for col in ["swimming_pool","week_price"]:
    index=df[df[col].isin(["-"])].index
    df.drop(index, inplace=True)
```

Voy a normalizar los datos de pets_allowed para que haya solo yes/no y luego lo codificaré a 1/0

```python
df["pets_allowed"] = df["pets_allowed"].replace({"pets allowed": "yes", "never": "no","we love pets":"yes"})
df["pets_allowed"] = df["pets_allowed"].replace({"yes":1,"no":0}).astype("int")
```

Voy a eliminar las filas para los valores nulos que aparecían para name y host_name

```python
df.dropna(subset=["name","host_name"],inplace=True)
#Como es posible que haya sitios que no tengan review voy
#a cambiar los vacios por 0, al igual que la fecha de la última review
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
df["last_review"] = df["last_review"].fillna("1900-01-01")
```

Ahora voy a convertir algunas columnas a tipo de datos específicos

```python
df["last_review"] = pd.to_datetime(df["last_review"])
df["swimming_pool"] = pd.to_numeric(df["swimming_pool"])
df["week_price"] = pd.to_numeric(df["week_price"])
df["price"] = pd.to_numeric(df["price"])
```

Como se ve cuando he agrupado los valores de las carácterísticas hay valores que no concuerdan, así que voy a borrar las filas que no lo cumplan

```python
#Fuera de rango
df.drop(df[df["availability_365"]>365].index, inplace=True)
df.drop(df[df["minimum_nights"]<0].index, inplace=True)
df.drop(df[df["price"]<=0].index, inplace=True)
```

Ahora vamos a ver los outliers

```python
def detectar_outliers(df, columna):
    plt.title(f"Detección de outliers: {columna}")
    plt.boxplot(df[columna])
    plt.show()
    print(df[columna].describe())

    # Calcular límites
    q1 = df[columna].quantile(0.25)
    q3 = df[columna].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    print(f"Límites para {columna}: {lower_limit}, {upper_limit}")

    # Puedo eliminar o escalar a los límites, si elimino borro demasiadas filas aprox
    #unas 12000 filas por lo que mejor escalo los datos.
    #Aunque también depende un poco de que se quira conseguir o buscar
    # Esto borra las filas df = df[(df[columna] >= lower_limit) & (df[columna] <= upper_limit)]
    df[columna] = np.clip(df[columna], lower_limit, upper_limit)

    plt.title(f"Detección de outliers: {columna}")
    plt.boxplot(df[columna])
    plt.show()
    print(df[columna].describe())
    print("--------------------------------------")
    return df

# Aplicar a columnas relevantes
outlier_columns = ["price", "week_price", "minimum_nights"]
for col in outlier_columns:
    df = detectar_outliers(df, col)
```

¿Qué podemos aprender sobre diferentes anfitriones y áreas?

Vemos que anfitriones tienen más propiedades y en que áreas hay más alojamientos

```python
barrios = df["neighbourhood"].value_counts().head(10)

barrios.plot(kind="bar", color="skyblue",edgecolor="black")

plt.title("Número de Alojamientos en los 10 Barrios con Más Alojamientos")
plt.xlabel("Barrios")
plt.ylabel("Número de Alojamientos")
plt.show()
```

```python
prop = df.groupby("host_id")["id"].count().sort_values(ascending=False).head(10)

prop.plot(kind="bar", color="skyblue",edgecolor="black")

plt.title("Número de Alojamientos de los 10 Anfitriones con Más Alojamientos")
plt.xlabel("Propietario ID")
plt.ylabel("Número de Alojamientos")
plt.show()
```

¿Qué podemos aprender de las predicciones? (por ejemplo:
ubicaciones, precios, reseñas, etc.) Podemos ver la media de precios en los 10 barrios con más alojamientos al igual que el propietario que más alojamientos tiene es el que más reseñas ha recibido.

```python
media_precio_barrios = df[df["neighbourhood"].isin(barrios.index)].groupby("neighbourhood")["price"].mean()

media_precio_barrios.sort_values(ascending=False).plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Promedio de Precios en los 10 Barrios con Más Alojamientos")
plt.xlabel("Barrios")
plt.ylabel("Precio Medio")
plt.show()
```

```python
total_review_prop = df[df["host_id"].isin(prop.index)].groupby("host_id")["number_of_reviews"].sum()

total_review_prop.sort_values(ascending=False).plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Total de Reseñas según los 10 Anfitriones con Más Alojamientos")
plt.xlabel("Host ID")
plt.ylabel("Total de Reseñas")
plt.show()
```
